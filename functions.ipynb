# Functions in python

- Reusable block of code that performs a task.

## Syntax

```python
def function_name(params):
    # Function body
    return some_value
```

- Where params and return are optional.

````python
# Function definition
def add(num1, num2):
    return num1 + num2

# Function call
result = add(1, 4)

print(result)
````

## Positional arguments

- The order of arguments matter.
- Python matches arguments by position.

````python
def divide(a, b):
    print(a / b);

divide(10, 5) 
````

````python
divide(5, 10)
````

## Keyword arguments

- We can also explicitly name params.
- Useful when function has many parameters.

````python
def introduce(name, age, city):
    print(name, age, city) 
````

````python
introduce(name="parth", age=28, city="mumbai")
````

## Mixing positional and keywords arguments

- Positional arguments have to come before keyword arguments.

````python
def create_user(name, age, city):
    print(name, age, city)

create_user("Alex", age=25, city="Mumbai")
````

````python
# The below will throw and error because of the order of positional argument

create_user(name="Alex", 25, "London")
````

## Local variables

- Variables created inside a function are normally local to that function. 

````python
def calculate():
    x = 10
    y = 20
    return x + y
````

````python
print(x)
````

- This is because `x` exists only inside `calculate()`.

## Global variables

- Variables defined outside function exist in an outer or global scope.

````python
tax_rate = 0.20

def calculate_tax(price):
    return price * tax_rate
````

````python
print(tax_rate)
````

- In the above code, if the value of `tax_rate` is not found within the function `calculate_tax`, python will look outside its scope.

````python
test()
print(x)
````

## LEGB scope rule

- Python resolves variable names using LEGB rule as follows:
    - L = local
    - E = enclosing
    - G = global
    - B = built-in

````python
x = "global"

def outer():
    x = "enclosing"

    def inner():
        print(x)

    inner()

outer()
````

- Python first searches local -> enclosing -> global -> built-in
- Built-in includes `print`, `len`, `sum`, `max`, `min`, `list`, `dict`.
- This is the reason why doing `list = [1,2,3,4,5]` is a bad idea.

## `global` keyword

- You can explicitly modify a global variable using the `global` keyword.~

````python
count = 0

def increment():
    global count
    count += 1
````

````python
increment()
increment()
print(count)
````

- We could have just done count += 1 but incase there was any other variable with the same name as count, using the global keyword targets the global count variable.
- However, avoid global keyword unless there is a strong reason.

## `nonlocal` keyword

- Widely works with nested function.

````python
def outer():
    count = 1

    def increment():
        nonlocal count
        count += 1

    increment()
    increment()
    print(count)

outer()
````

- The nonlocal keyword makes the variable belong to an enclosing function scope rather than the local scope.
- It is less commonly used in everyday's code but is more important when understanding closures.

## Mutable vs Immutable arguments

````python
def change_number(x):
    x = 100

number = 5
change_number(number)

print(number)

# This means pass by value right?
````

````python
def add_items(items):
    items.append("apple")

fruits = []
add_items(fruits)

print(fruits)

# This means pass by referrence right?
````

- This means the function received a reference to the same list object.
- Mutating that list affects the original object.
- Common mutable objects are as follows:
    - list
    - dict
    - set
    - most class instances
- Common immutable objects are as follows:
    - int
    - float
    - str
    - bool
    - tuple
    - frozenset

- A rule to know is that python widely uses object references.

## Reassignment vs mutation 

````python
# Reassignment

def change(items):
    items = [100, 200]

numbers = [1, 2, 3, 4, 5]
change(numbers)
print(numbers)
````

- The local variable `items` is pointing somewhere else. 

````python
# Mutation

def change(items):
    items.append(100)

numbers = [1, 2, 3]
change(numbers)

print(numbers)
````

## The mutable default argument trap

- Never casually write:

````python
def add_student(name, students = []):
    students.append(name)
    return students

add_student("parth")
````

````python
add_student("jay")
````

- The expectation was `["jay"]` but we got `["parth", "jay"]`.
- Default args values are created once the function is defined and not everytime when its called.

````python
# Correct way, classic python interview question.
# Need to understand it more clearly.

def add_student(name, students=None):
    if students is None:
        students = []
    students.append(name)
    return students
````

````python
add_student("parth")
````

````python
add_student("jay")
````

## `*args`

- Sometimes we do not know how many positional arguments are required.
- Hence, we use the below pattern:

````python
def add_numbers(*numbers):
    return sum(numbers)
````

````python
add_numbers(1, 2, 3)
````

````python
add_numbers(1, 2, 3, 4, 5)
````

- Inside the function, `numbers` is tuple.
- The conventional name is `*args` but the name isn't special.
- `*values` works too.
- The `*` is more important.

## `**kwargs`

- It captures arbitary keyword arguments.

````python
def create_profile(**kwargs):
    print(kwargs)

create_profile(name="parth", age=28, city="London")
````

- Inside `**kwargs`:

{
    "name": "Alex",
    "age": 25,
    "city": "London"
}

- `*args` = tuple of positional arguments.
- `**kwargs` = dictionary of keyword arguments.

## Docstring

````python
def calculate_average(numbers):
    """Return the arithmetic mean of a sequence of numbers."""
    return sum(numbers) / len(numbers)

# The line with tripled quoted string just below the definition is the docstring.
````

````python
# You can inspect it as follows:

calculate_average.__doc__
````

````python
# A better docstring is as follows:

def calculate_discount(price: float, discount: float) -> float:
    """
    Calculate the final price after applying a percentage discount.

    Args:
        price: Original product price.
        discount: Discount percentage.

    Returns:
        Final price after discount.
    """
    return price * (1 - discount / 100)

# However, do not write enormously large docstrings for very obvious functions.
````

## Passing functions into functions

````python
def add(a, b):
    return a + b 
````

````python
def calculate(operation, x, y):
    return operation(x, y)
````

````python
calculate(add, 10, 20)
````

- We are passing the `add` function as an argument to `calculate`. 
- Notice that we are not calling `add()`.
- So passing a function as an argument to another function makes the another function as higher-order function. 

## Returning functions from functions

````python
def get_operation():
    def add(a, b):
        return a + b

    return add
````

````python
operation = get_operation()

print(operation(10, 20))
````

- The above might feel strange but it is the foundation of decorators.
