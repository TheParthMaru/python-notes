# Functions in python

- Reusable block of code that performs a task.

## Syntax

```python
def function_name(params):
    # Function body
    return some_value
```

- Where params and return are optional.

````python
# Function definition
def add(num1, num2):
    return num1 + num2

# Function call
result = add(1, 4)

print(result)
````

## Positional arguments

- The order of arguments matter.
- Python matches arguments by position.

````python
def divide(a, b):
    print(a / b);

divide(10, 5) 
````

````python
divide(5, 10)
````

## Keyword arguments

- We can also explicitly name params.
- Useful when function has many parameters.

````python
def introduce(name, age, city):
    print(name, age, city) 
````

````python
introduce(name="parth", age=28, city="mumbai")
````

````python
# The order does not matter when using keyword arguments
introduce(age=24, city="leicester", name="jay")
````

## Mixing positional and keywords arguments

- Positional arguments have to come before keyword arguments.

````python
def create_user(name, age, city):
    print(name, age, city)

create_user("Alex", age=25, city="Mumbai")
````

````python
# The below will throw and error because of the order of positional argument

create_user(name="Alex", 25, "London")
````

## Return statement

- Mostly same as in every other language.
- By default a function returns `None` even if there is no return statement mentioned.

## Returning multiple values

````python
def calculate(a, b):
    return a+b, a-b
````

````python
# A tuple is returned in the above case

result = calculate(5, 10)

print(result)
````

````python
# We can also unpack it as follows:

addition, substraction = calculate(10, 20)

print(addition)
print(substraction)
````

## Default parameters

- We can provide a default value to parameters as follows:

````python
def greet(name, greeting="Hello"):
    print(f"{greeting} {name}") 
````

````python
greet("parth")
````

````python
greet("parth", "good morning")
````
