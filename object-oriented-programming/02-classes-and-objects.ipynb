# Classes and Objects

- Class is a blueprint/design/custom type you define.
- An object is a real instance created from that class.
- Without classes, we may store everything using dictionaries which is okay for small simple data.
- But as the data grows, a class becomes more useful.
- When you are creating your own class, you are creating your own type.

## Syntax

```python
# Creating a customer class
class Customer:
    pass
```

```python
# Creating a object of class customer
customer_1 = Customer()
customer_2 = Customer()
```

```python
print(type(customer_1))
```

```python
# Building a product class

class Product:
    pass

# Creating its objects
product_1 = Product()
product_2 = Product()
```

```python
print(product_1)
print(product_2)
```

- The above verifies that both the objects are different because they have different memory addresses.

```python
# Adding attributes
product_1.name = "keyboard"
product_1.price = 2000.00
```

```python
product_2.name = "mouse"
product_2.price = 150.00
```

- Classes give concept.
- Objects holds the actual values.

## Important naming conventions

- For classes, use pascal case.
- For objects, use snake case.

## Attributes

- It is data attached to the object or class.
- In simpler terms, attribute is the information stored inside an object.

```python
class Product:
    pass

product = Product()

product.name = "keyboard"
product.price = 50.00
product.category = "accessories"
```

- Here, `name`, `price` and `category` are attributes of the `product` object.
- So the object now carries its own data.
- Python allows you to add attributes dynamically.
- Use snake case for attributes name.

## Methods

- It is a function that belongs to a class or object.
- Attributes store data.
- Methods define behaviour.

```python
class Product:
    def display_summary(self):
        return "This is a product"
```

```python
product = Product()

product.display_summary()
```

- A function exist independently whereas the method belongs to a class/object.
- Use verbs and snake case for naming convention.

## The `self` parameter

- It refers to the current object.
- When you define a method inside a class, Python expects the first parameter to represent the object that is calling the method.

```python
class Customer:
    def display_profile(self):
        return "Customer profile"
```

- Here, `self` = the object that called `display_profile()`.
- That means the object of `Customer` gets passed into the method automatically.
- Without `self`, a method would not know which object’s data to use.
- It is same like `this` keyword in Java.

```python
class Product:
    def display_product_details(self):
        print(self.name, self.price)
```

```python
product = Product()
product.name = "samsung phone"
product.price = 50000.00
```

```python
product.display_product_details()
```

- Python also automatically passes the object as the first argument.
- `self` is not a keyword in python.
- It is a convention. Instead of `self`, you can also use `this` which will work same as `self`.
- This happens because the first parameter is considered as object reference.

```python
class Transaction:
    def show_transaction_details(this):
        print(this.value, this.status)
```

```python
transaction_1 = Transaction()
transaction_1.value = 1000.00
transaction_1.status = "pending"

transaction_1.show_transaction_details()
```

- As proved above, `self` is a just a convention.
- That's what python people call it.
- Professional python developers expect self.
