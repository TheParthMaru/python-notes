# Types of variables

## Instance variables

- An instance variable is an attribute that belongs to a specific object.
- In Python, an object is also called an instance of a class.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
```

- Each `Product` gets its own `name` and `price`.
- Instance variables allow different objects from the same class to store different data.
