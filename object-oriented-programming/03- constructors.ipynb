# Constructor

- `__init__` is a special method that runs automatically when you create an object.
- It is used to initialise the object with starting data.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
```

```python
product_1 = Product("keyboard", 49.99)
```

- Earlier we were manually adding attributes which makes the code messy.
- If we manually add attributes, we could forget certain attributes and create inconsistent objects.
- With `__init__`, we force every product to start with the required data.
- In real projects, `__init__` helps you create objects in a predictable state.
- We can also provide default values as follows:

```python
class Customer:
    def __init__(self, name, email, is_active=True):
        self.name = name
        self.email = email
        self.is_active = is_active
```

## Basic validations in `__init__`

- We can also validate data when the object is created.

```python
class Transaction:
    def __init__(self, amount, category):
        if amount <= 0:
            raise ValueError("Amount must be greater than 0")

        self.amount = amount
        self.category = category
```

- This is good because it prevents invalid objects from being created.
- In real data work, validation is extremely important. 
- You do not want invalid transactions, missing customer IDs, or broken file paths spreading through your program.
