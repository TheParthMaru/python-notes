# Intro to Object Oriented Programming

- OOP means writing code by modelling things as objects.
- An object usually contains data + behaviour.
- In Python, we say attributes + methods.
- For example, a `Product` object may have:
    - Attributes: id, name, price, description, category
    - Methods: `apply_discount()`, `update_price()`, `is_in_stock()`, `display_summary()`
- OOP enables us to group related data and behaviour together.

## Why does OOP matter?

- In real projects, the code grows quickly.
- For a tiny script, procedural code can be fine.
- But when you have loads of products, 1000s of transactions, multiple customers, multiple files, validation rules, cleaning rules, DB operations, reports, API calls, etc, procedural code becomes very messy.
- OOP helps you create a clear structure like:
    - Product
    - Customer
    - Transaction
    - CSVProcessor
    - DataCleaner
    - DataValidator
    - ETLPipeline
    - ReportGenerator
- Each class has a clear responsibility.
- This is called as separation of concerns.
- For example:
    - CSVProcessor -> Responsible for reading and processing CSV files.
    - DataCleaner -> Responsible for cleaning data.
    - DataValidator -> Responsible for checking quality of data.
    - ReportGenerator -> Responsible for generating reports.
- This is much better than putting everything in a single large python file.
