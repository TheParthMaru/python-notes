# Exception Handling

## Understanding errors and exceptions

- When a program executes a sequence of operations, normally everything follows the expected path.
- But sometimes an operation cannot successfully complete.
- Some of those operations are:
    - divide by 0
    - opening a file that does not exist
    - coverting "hello" to an integer
- When something like above occurs, an error is thrown and the normal flow of the code is interrupted and stopped.
- That is why we require exception handling.
- An exception is a structured signal that interrupts the normal execution path because something exceptional happened.

```python
# For example

num = int("hello")
```

- Python cannot produce a meaningful integer.
- Instead of inventing a value such as 0, python raises an exception called `ValueError`.
