# Exception Handling

## Understanding errors and exceptions

- When a program executes a sequence of operations, normally everything follows the expected path.
- But sometimes an operation cannot successfully complete.
- Some of those operations are:
    - divide by 0
    - opening a file that does not exist
    - coverting "hello" to an integer
- When something like above occurs, an error is thrown and the normal flow of the code is interrupted and stopped.
- That is why we require exception handling.
- An exception is a structured signal that interrupts the normal execution path because something exceptional happened.

````python
# For example

num = int("hello")
````

- Python cannot produce a meaningful integer.
- Instead of inventing a value such as 0, python raises an exception called `ValueError`.

## Why do exception handling mechanisms exist?

````python
# Assume the below function

def load_user(user_id):
    ...
````

- If the requested user cannot be loaded, we have have the possible approaches as:
    - return None OR 
    - return False OR
    - return -1 OR
    - raise UserNotFoundException(...)

- Exceptions gives us a standard way to separate normal result from operation failed.
- Exceptions solves ambiguity.

## Syntax errors vs runtime exceptions

### Syntax errors

- Syntax error means that the programming language cannot even understand the program's structure.
- For many languages, it could be a missing `;`, for python it could be a missing `:`.
- Python would report a `SyntaxError` which means that the program isn't a valid python code.

### Runtime exception

- Here, the code is syntactically valid, but something goes wrong while it is running.
- For example,

```py
age = int("twenty")
```

- The above code has perfectly valid python syntax.
- But during the execution, it would fail as `twenty` cannot be converted to an `integer`.
- Python will raise `ValueError`.

### Summary

- Syntax problem = python cannot properly execute the program.
- Runtime exception = python started executing, but an operation failed.

## Why exception handling is important?

````python
# Assume the below code

print("start")

number = int("hello")

print("end")
````

- If we look at the output, `start` gets printed, then we get an error, but `end` is never printed.
- We get the traceback as above where we get to know approximately the following:
    - where execution was
    - which functions were called
    - where the failure occurred
    - what exception occurred
- The execution conceptually looks as follows:

```bash
print("Start")
      ↓
int("hello")
      ↓
ValueError raised
      ↓
normal execution stops
      ↓
Python searches for an exception handler
      ↓
none found
      ↓
program terminates with traceback
```

## `try` and `except`

- Sometimes you know an operation might fail, and you know what your program should do when it fails.
- For that very reason, python allows us to handle the exceptions gracefully as follows:

### Syntax

```py
try:
    risky_opeartion()
except SomeException:
    handle_failure()
```

````python
# Example

print("start")

try:
    num = int("hello")
except ValueError:
    print("That was not a valid integer")

print("end")
````

- Now the program doesn't crash and throw a stack trace of errors.
- Also, the program finishes as expected executing all necessary code.
- Conceptually:

```bash
try operation
   ↓
works? ── yes ──→ continue normally
   │
   no
   ↓
matching except block
   ↓
continue execution
```

- The exception changes the execution path when handled properly.

## Catching specific exception types

````python
try:
    print(10/0)
except ValueError:
    print("Invalid division")
````

- We cannot use random exceptions anywhere.
